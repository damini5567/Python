## STRINGS

##### 1) Single quotes - ' '


```python
text='Python is a programming language.'
print(text)
print(type(text))
```

##### 2) Double quotes - " "

```python
text1="Python is a programming language."
print(text1)
print(type(text1))
```

##### 3) Tripple quotes - '''  '''    or    """  """

###### i) single line string

```python
text3='''Python is a programming language.'''
print(text3)
print(type(text3))
```

```python
text4="""Python is a programming language."""
print(text4)
print(type(text4))
```

###### ii) Multiline String

```python
text5='''Hello 
welcome to AI Course
'''
print(text5)
print(type(text5))
```

```python
text6="""Hello 
welcome to AI Course
"""
print(text6)
```
